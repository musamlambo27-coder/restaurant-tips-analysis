# Tips Analysis

This notebook contains the analysis from `tips_analysis.py` using the seaborn `tips` dataset.

```python
import seaborn as sns
import pandas as pd 

df = sns.load_dataset('tips')
df.head()

avg_by_day = df.groupby('day')['total_bill'].mean().sort_values(ascending=False)
print(avg_by_day)

df['tip_percentage'] = (df['tip'] / df['total_bill']) * 100
print(df.groupby('day')['tip_percentage'].mean())

import matplotlib.pyplot as plt
avg_by_day.plot(kind='bar', figsize=(8,5), title='Average Total Bill by Day', color='teal')
plt.ylabel('Average Total Bill')
plt.tight_layout()
plt.savefig('average_total_bill_by_day.png')
plt.show()
```
